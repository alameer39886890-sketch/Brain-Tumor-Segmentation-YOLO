# Brain Tumor Segmentation with YOLO11

This notebook replaces the custom TensorFlow CNN with **YOLO11 segmentation**.
The Kaggle dataset contains images and binary masks, so the masks are converted to YOLO polygon labels automatically.

````python
import os
import random
import shutil
from pathlib import Path

import cv2
import numpy as np
import yaml
import matplotlib.pyplot as plt
from ultralytics import YOLO

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
````

````python
# Download the original Kaggle dataset
import kagglehub

path = kagglehub.dataset_download("nikhilroxtomar/brain-tumor-segmentation")
Image_dir = Path(path) / "images"
Mask_dir = Path(path) / "masks"

print("Dataset path:", path)
print("Images:", len(list(Image_dir.iterdir())))
print("Masks:", len(list(Mask_dir.iterdir())))
````

## Convert image/mask pairs to YOLO segmentation format

YOLO segmentation labels use this format per object:

```text
class_id x1 y1 x2 y2 ...
```

All polygon coordinates are normalized to `[0, 1]`. The code below extracts contours from each binary mask and creates one label file per image.


````python
# YOLO dataset folders
YOLO_ROOT = Path("brain_tumor_yolo")
for folder in [
    YOLO_ROOT / "images" / "train",
    YOLO_ROOT / "images" / "val",
    YOLO_ROOT / "labels" / "train",
    YOLO_ROOT / "labels" / "val",
]:
    folder.mkdir(parents=True, exist_ok=True)

# Match only images that have a corresponding mask
image_paths = sorted([
    p for p in Image_dir.iterdir()
    if p.suffix.lower() in {".jpg", ".jpeg", ".png"}
    and (Mask_dir / p.name).exists()
])

random.Random(SEED).shuffle(image_paths)
split_index = int(0.8 * len(image_paths))
train_images = image_paths[:split_index]
val_images = image_paths[split_index:]

print(f"Total pairs: {len(image_paths)}")
print(f"Train: {len(train_images)} | Validation: {len(val_images)}")
````

````python
def mask_to_yolo_segments(mask_path, min_area=10):
    # Convert a binary mask into YOLO normalized polygon rows.
    mask = cv2.imread(str(mask_path), cv2.IMREAD_GRAYSCALE)
    if mask is None:
        raise FileNotFoundError(f"Cannot read mask: {mask_path}")

    binary = (mask > 127).astype(np.uint8) * 255
    contours, _ = cv2.findContours(
        binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
    )

    height, width = binary.shape
    rows = []
    for contour in contours:
        if cv2.contourArea(contour) < min_area:
            continue

        # Polygon simplification keeps labels compact while preserving shape.
        epsilon = 0.002 * cv2.arcLength(contour, True)
        polygon = cv2.approxPolyDP(contour, epsilon, True).reshape(-1, 2)
        if len(polygon) < 3:
            continue

        points = []
        for x, y in polygon:
            points.extend([float(x) / width, float(y) / height])
        rows.append("0 " + " ".join(f"{value:.6f}" for value in points))

    return rows


def prepare_split(paths, split):
    for image_path in paths:
        destination_image = YOLO_ROOT / "images" / split / image_path.name
        destination_label = YOLO_ROOT / "labels" / split / f"{image_path.stem}.txt"
        shutil.copy2(image_path, destination_image)

        rows = mask_to_yolo_segments(Mask_dir / image_path.name)
        # Empty files are valid YOLO labels and mean that no object is present.
        destination_label.write_text("\n".join(rows), encoding="utf-8")

prepare_split(train_images, "train")
prepare_split(val_images, "val")
print("YOLO dataset created at:", YOLO_ROOT.resolve())
````

````python
# Create the YOLO data configuration
DATA_YAML = YOLO_ROOT / "data.yaml"
data_config = {
    "path": str(YOLO_ROOT.resolve()),
    "train": "images/train",
    "val": "images/val",
    "names": {0: "brain_tumor"},
}
DATA_YAML.write_text(yaml.safe_dump(data_config, sort_keys=False), encoding="utf-8")
print(DATA_YAML.read_text())
````

````python
# Optional: inspect one image and its converted mask polygon
sample = train_images[0]
image = cv2.cvtColor(cv2.imread(str(sample)), cv2.COLOR_BGR2RGB)
mask = cv2.imread(str(Mask_dir / sample.name), cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(mask, cmap="gray")
plt.title("Original mask")
plt.axis("off")
plt.show()
````

## Train YOLO11 segmentation

`yolo11n-seg.pt` is the small pretrained segmentation model. Unlike the original CNN, YOLO handles the segmentation head, augmentation, validation, and checkpointing.


````python
# Train YOLO11 segmentation
model = YOLO("yolo11n-seg.pt")

results = model.train(
    data=str(DATA_YAML),
    epochs=10,
    imgsz=640,
    batch=16,
    device=0,       # change to "cpu" if no GPU is available
    workers=2,
    project="runs",
    name="brain_tumor_yolo11_seg",
    exist_ok=True,
    pretrained=True,
)
````

## Important notes

- The original `accuracy` metric is not a useful replacement for segmentation quality because tumor pixels are usually a small part of the image. YOLO reports mask mAP and related metrics.
- If GPU memory is insufficient, reduce `batch=16` to `batch=8` or `batch=4`.
- If training on CPU, change both `device=0` values to `device="cpu"`.
- The trained model is saved at `runs/brain_tumor_yolo11_seg/weights/best.pt`.
